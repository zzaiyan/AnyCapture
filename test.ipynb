```python
from anycapture import get_local
get_local.activate()
```

```python
import torch
import torch.nn as nn

class SimpleModel(nn.Module):
    def __init__(self):
        super(SimpleModel, self).__init__()
        self.linear = nn.Linear(10, 1)

    @get_local('x', 'out')
    def forward(self, x):
        out = self.linear(x)
        return out
```

```python
dummy_input = torch.randn(5, 10)
model = SimpleModel()
with torch.no_grad():
    output = model(dummy_input)
```
